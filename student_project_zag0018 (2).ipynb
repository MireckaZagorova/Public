## 🧹 3) Preprocessing pipeline

Vybereme pouze relevantní sloupce pro analýzu.

```
selected_columns = [
    "age","Medu","Fedu","traveltime","studytime","failures","famrel","freetime",
    "goout","Dalc","Walc","health","absences",
    "sex","address","famsize","Pstatus","schoolsup","famsup","paid",
    "activities","nursery","higher","internet","romantic"
]

X = df[selected_columns]
X.head()
```

Následně převedeme kategorické proměnné (např. schoolsup, famsup, paid, activities a další) na číselné hodnoty pomocí metody One-hot Encoding, která z nich vytvoří binární sloupce s hodnotami True/False (později reprezentované jako 0/1).

```
X_encoded = pd.get_dummies(X, drop_first=True)
X_encoded.head()
```

Nakonec všechny proměnné standardizujeme, aby měly stejnou škálu – získáme tak číselnou matici vhodnou pro clustering.

```
import pandas as pd
from sklearn.preprocessing import StandardScaler


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)

X_scaled_df = pd.DataFrame(X_scaled, columns=X_encoded.columns)
X_scaled_df.head()

```

## 🌀 4) K-Means clustering

### 4.1. Elbow metoda - nalezení optimálního počtu clusterů


```
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Zkusíme počty clusterů 1 až 10
inertia_values = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled_df)
    inertia_values.append(kmeans.inertia_)

# Vykreslení Elbow grafu
plt.figure(figsize=(8,5))
plt.plot(K_range, inertia_values, marker='o')
plt.xlabel("Počet clusterů (k)")
plt.ylabel("Inertia")
plt.title("Elbow metoda")
plt.grid(True)
plt.show();
```

Graf Inertie ukazuje, jak se mění kompaktnost clusterů při různém počtu k. Největší pokles Inertie nastává mezi 1–3 clustery, poté se křivka zpomaluje. Zlom („loket“) je proto kolem hodnoty k = 3, případně mírně posunutý ke k = 4. Od k ≥ 5 už další přidávání clusterů nepřináší významné zlepšení. Metoda nám tedy nazančuje, že vhodná počet clusterů je k = 3. Ověříme to metodou Silhouette Score

### 4.2. Metoda Silhouette Score

```
from sklearn.metrics import silhouette_score

scores = {}
for k in range(2, 11):
    km = KMeans(n_clusters=k, random_state=42)
    labels = km.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    scores[k] = score;

scores
```

Zde vidíme, že nejlepší - nejvyšší - silhouette má k = 2.
Dále přesto budeme pokračovat s k = 3.

#### 4.3.1. Vytvoření clusteru

Na předchozím scatterplotu lze jasně vidět rozdíly mezi clustery z hlediska studijních výsledků.
Nyní se pokusíme vizualizovat vztah mezi počtem absencí a počtem propadnutí.

```
sns.scatterplot(
    data=df,
    x="absences",
    y="failures",
    hue="cluster",
    palette=[ "#3498db", "#2ecc71","#c2185b"],
    alpha=0.85,
    s=70
);

plt.xlabel("Počet absencí")
plt.ylabel("Počet propadnutí")
plt.title("Clustery podle absence a počtu propadnutí");
```

Na tomto scatterplotu je patrné, že body se ve velké míře překrývají, což snižuje přehlednost vizualizace.
Proto se v dalším kroku pokusíme odfiltrovat extrémní hodnoty, aby byl graf čitelnější.

#### 4.4.3. PCA vizualizace

Pomocí PCA snížíme datovou dimenzi tak, aby bylo možné zobrazit jednotlivé clustery ve dvourozměrném prostoru. PCA slouží pouze jako vizualizační nástroj – neukazuje přesné vztahy mezi proměnnými, ale umožňuje přehledně vidět, jak jsou clustery od sebe oddělené.

```
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

df["PCA1"] = X_pca[:, 0]
df["PCA2"] = X_pca[:, 1]

sns.scatterplot(
    data=df,
    x="PCA1",
    y="PCA2",
    hue="cluster",
    palette=["#1f77b4", "#2ca02c", "#d62728"],  # modrá, zelená, růžová
    alpha=0.8
)

plt.title("Vizualizace clusterů pomocí PCA")
plt.show();
```

Na PCA vizualizaci vidíme, že cluster 1 (zelené body) tvoří poměrně kompaktní a soudržnou skupinu. To odpovídá tomu, že jde o studenty s nejlepšími výsledky a stabilním chováním.

Cluster 0 (modré body) je rozptýlenější, ale stále se drží poblíž clusteru 1. To zapadá do profilu průměrných studentů – nejsou extrémně slabí ani extrémně dobří, proto se jejich hodnoty překrývají s oběma směry.

Cluster 2 (červené body) je nejvíce rozptýlený a nemá jasné jádro. To ukazuje, že jde o velmi různorodou skupinu studentů, u kterých se kombinují problémy jako horší známky, častější absence nebo předchozí propadnutí. Je to tedy nejméně stabilní a nejslabší cluster.

## 🌳 5) Agglomerative clustering

Druhou metodou, kterou jsem vyzkoušela, je Agglomerative clustering. Ten vytváří skupiny tak, že začíná od jednotlivců a pomalu je spojuje dohromady.
Opět jsem zvolila 3 clustery, aby se to dalo srovnat s K-Means. Výsledek jsem pak vizualizovala a porovnala s první metodou.

```
from sklearn.cluster import AgglomerativeClustering

agg = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

df["cluster_agg"] = agg.fit_predict(X_scaled)
```

```
sns.scatterplot(
    x=df["PCA1"],
    y=df["PCA2"],
    hue=df["cluster_agg"],
    palette=["#3498db", "#2ecc71", "#c2185b"],
    alpha=0.8
)

plt.title("Vizualizace Agglomerative clusteringu pomocí PCA")
plt.show();
```

Z grafu je patrné, že Agglomerative clustering vytváří shluky odlišným způsobem než K-Means. Cluster 1 (zelený) tvoří kompaktní skupinu podobně jako u K-Means, což potvrzuje, že se jedná o nejlepší studenty. Naopak cluster 2 (červený) je rozprostřený po širší oblasti PCA prostoru, což odpovídá různorodým problémům těchto studentů. Cluster 0 (modrý) zůstává mezi ostatními skupinami a představuje průměrné studenty. Výsledky tedy potvrzují stejný trend jako K-Means, ale hranice mezi clustery jsou mírně odlišné.

### 6.1. Vyhodnocení hypotéz

* Čím vyšší počet absencí, tím horší budou studijní výsledky (G3).

* Studenti, kteří věnují více času studiu (studytime), budou mít lepší výsledky.

* Studenti, kteří více chodí ven, častěji pijí alkohol nebo jsou ve vztahu, budou mít horší studijní výkon.

* Čím více propadnutí studenta, tím má horší studijní výsledky.

První tři hypotézy se nám podařilo vyvrátit.
Z dat vyplynulo, že:

Absence mají pouze velmi slabou korelaci se známkou G3 a žádný jasný vztah se neobjevil ani v grafech.

Čas věnovaný studiu (studytime) nemá téměř žádný dopad na výslednou známku G3 – studenti s dobrými i slabými výsledky studují podobně dlouho.

U volnočasových aktivit, jako je chození ven, pití alkoholu, trávení času s přáteli nebo vztah, se taktéž neprojevil žádný silný vliv na studijní výkon.

Naopak čtvrtá hypotéza se potvrdila.
Počet propadnutí má jednoznačnou negativní vazbu na výsledky – studenti, kteří alespoň jednou propadli, vykazují výrazně horší závěrečnou známku G3.

Z EDA bylo patrné, že největší vliv na známky mají předchozí výsledky a počet propadnutí. Naopak absence nebo množství času věnovaného studiu nemají výraznou vazbu na závěrečné známky.

 * velmi úspěšní studenti (nejlepší výsledky, vysoká podpora rodiny, minimum problémů),

 * rizikoví studenti (nejhorší výsledky, časté propadnutí, vyšší absence a horší zdraví).

Naopak průměrná a problematická skupina studentů se mezi metodami více míchá, což potvrzuje, že tyto dvě skupiny nejsou tak jednoznačně oddělitelné.

Celkově obě metody potvrdily podobnou strukturu dat a ukázaly, že studijní výsledky studentů nejsou náhodné — vytvářejí tři hlavní typy studentů s poměrně jasnými charakteristikami. K-Means poskytl přehlednější a stabilnější segmentaci, zatímco Agglomerative clustering nabídl alternativní pohled a ukázal, že některé skupiny studentů nejsou příliš homogenní.
